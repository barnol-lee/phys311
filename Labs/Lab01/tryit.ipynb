```python
import numpy as np
import matplotlib.pyplot as plt

# Physical parameters
g = 9.81          # m/s², gravitational acceleration
m = 0.145         # mass of a baseball, kg
r = 0.0366        # radius of a baseball, m
Cd = 0.47         # drag coefficient for a sphere
rho = 1.225       # air density, kg/m³
A = np.pi * r**2  # cross-sectional areag = 9.81        
dt = 0.01         # time step in seconds

# Combine the drag constants into one number
b = 0.5 * Cd * rho * A  # so F_drag = -b * |v| * v

# Initial conditions: launched at 45°, speed 20 m/s
v0 = 40.0
angle = np.radians(45)
x, y = 0.0, 0.0
vx, vy = v0 * np.cos(angle), v0 * np.sin(angle)

# Storage lists for plotting
xs_drag, ys_drag = [x], [y]

# Euler integration loop
t = 0.0
while y >= 0.0:
    # Speed (magnitude of velocity)
    speed = np.sqrt(vx**2 + vy**2)

    # Forces: gravity + drag
    Fx = -(b / m) * speed * vx
    Fy = -g - (b / m) * speed * vy

    # Euler update
    vx += Fx * dt
    vy += Fy * dt
    x += vx * dt
    y += vy * dt

    t += dt
    xs_drag.append(x)
    ys_drag.append(y)

# Plot trajectory
plt.figure(figsize=(8, 4))
plt.plot(xs_drag, ys_drag, 'b-', linewidth=2)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Projectile Trajectory — Euler Method')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
g = 9.81         # m/s², gravitational acceleration
dt = 0.5        # time step in seconds
t_max = 4.0      # total simulation time

# Initial conditions: launched at 45°, speed 20 m/s
v0 = 20.0
angle = np.radians(45)
x, y = 0.0, 0.0
vx, vy = v0 * np.cos(angle), v0 * np.sin(angle)

# Storage lists for plotting
xs, ys, ts = [x], [y], [0.0]

# Euler integration loop
t = 0.0
while t < t_max and y >= 0.0:
    # Compute acceleration (only gravity here)
    ax = 0.0
    ay = -g

    # Update velocity (Euler)
    vx = vx + ax * dt
    vy = vy + ay * dt

    # Update position (Euler)
    x = x + vx * dt
    y = y + vy * dt

    t += dt
    xs.append(x)
    ys.append(y)
    ts.append(t)

# Plot trajectory
plt.figure(figsize=(8, 4))
plt.plot(xs, ys, 'b-', linewidth=2)
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title('Projectile Trajectory — Euler Method')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()
```
